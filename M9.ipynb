```python
#soal 1
waktu = 10
while waktu > 0:
  print(f"Sisa waktu: {waktu} detik")
  waktu -= 1
print("Selesai!")
```

```python
#soal 3
total_berat = 0
while True:
  total_berat += 5
  print(f"Total berat saat ini: {total_berat} kg")
  if total_berat >= 25:
    print("Kapasitas maksimum tercapai, menghentikan loop!")
    break
```
